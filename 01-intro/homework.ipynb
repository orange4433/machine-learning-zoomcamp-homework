```python
df1 = df[
    df['origin'] == 'Asia'
]
df1
```

```python
df2 = df1[['vehicle_weight','model_year']]
df2
```

```python
df3 = df2.iloc[[0,1,2,3,4,5,6]]
df3
```

```python
X = df3.values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
import numpy as np
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX_inv
```

```python
y = np.array([1100,1300,800,900,1000,1100,1200])
```

```python
y
```

```python
w = XTX_inv.dot(X.T.dot(y))
```
